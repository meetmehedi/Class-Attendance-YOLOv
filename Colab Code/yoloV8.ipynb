```
import os

# Path to the directory containing label files
label_dir = '/content/drive/MyDrive/DIU_Class_Student/Class_student_dataset/train/labels'

# Function to read label files and check for negative values
def check_negative_labels(label_dir):
    negative_labels = {}
    for label_file in os.listdir(label_dir):
        file_path = os.path.join(label_dir, label_file)
        with open(file_path, 'r') as f:
            lines = f.readlines()
            for line in lines:
                # Split the line into values
                values = line.strip().split()
                # Assuming the first value is the class ID
                if len(values) > 0:  # Check if the line is not empty
                    try:
                        label = int(values[0])  # Convert the first value to an integer
                        if label < 0:
                            if label_file not in negative_labels:
                                negative_labels[label_file] = []
                            negative_labels[label_file].append(label)
                    except ValueError:
                        print(f"Skipping line with invalid class ID in {label_file}: {line.strip()}")
    return negative_labels

# Check for negative labels
negative_labels = check_negative_labels(label_dir)

# Print the files and labels with negative values
print("Negative labels found:")
for file, labels in negative_labels.items():
    print(f"{file}: {labels}")
```

```
import os

# Path to the directory containing label files
label_dir = '/content/drive/MyDrive/DIU_Class_Student/Class_student_dataset/train/labels'

# Function to fix negative labels
def fix_negative_labels(label_dir):
    for label_file in os.listdir(label_dir):
        file_path = os.path.join(label_dir, label_file)
        with open(file_path, 'r') as f:
            lines = f.readlines()

        # Fix negative labels and rewrite the file
        with open(file_path, 'w') as f:
            for line in lines:
                values = line.strip().split() # Split the line into values
                if len(values) > 0: # Check if the line is not empty
                    try:
                        label = int(values[0])  # Convert the first value to an integer
                        if label < 0:
                            label = 0  # Replace negative labels with 0
                        f.write(f"{label} {' '.join(values[1:])}\n") # Rewrite the line with the corrected label
                    except ValueError:
                        print(f"Skipping line with invalid class ID in {label_file}: {line.strip()}")

# Fix negative labels
fix_negative_labels(label_dir)

# Check again to ensure no negative labels remain
negative_labels = check_negative_labels(label_dir) # Assuming check_negative_labels is defined elsewhere
print("After fixing, negative labels found:")
print(negative_labels)
```

```
import cv2
import os

# Path to the directory containing image files
image_dir = '/content/drive/MyDrive/DIU_Class_Student/Class_student_dataset/train/images'

# Function to check image files
def check_images(image_dir):
    corrupt_images = []
    for img_file in os.listdir(image_dir):
        img_path = os.path.join(image_dir, img_file)
        try:
            img = cv2.imread(img_path)
            if img is None:
                corrupt_images.append(img_path)
        except Exception as e:
            corrupt_images.append(img_path)
    return corrupt_images

# Check images
corrupt_images = check_images(image_dir)

# Print the corrupt images
print("Corrupt images found:")
for img_path in corrupt_images:
    print(img_path)
```

```
import os

# লেবেল ফাইলগুলির পাথ
label_dir = "/content/drive/MyDrive/DIU_Class_Student/Class_student_dataset/val/labels"

# লেবেল ফাইলগুলোতে কোঅর্ডিনেটস চেক এবং সংশোধন করা
for label_file in os.listdir(label_dir):
    label_path = os.path.join(label_dir, label_file)
    with open(label_path, 'r') as file:
        lines = file.readlines()
    new_lines = []
    for line in lines:
        # Skip lines that don't start with a number or contain non-numeric values after the first element
        if line[0].isdigit() and all(val.replace('.', '', 1).isdigit() for val in line.split()[1:]):
            values = list(map(float, line.split()))
            if all(0 <= val <= 1 for val in values[1:]):
                new_lines.append(line)
    with open(label_path, 'w') as file:
        file.writelines(new_lines)
```

```
import os

# ডেটাসেটের ফোল্ডারের পথ
train_label_path = "/content/drive/MyDrive/DIU_Class_Student/Class_student_dataset/train/labels"
val_label_path = "/content/drive/MyDrive/DIU_Class_Student/Class_student_dataset/val/labels"

def check_and_fix_labels(label_dir):
    """লেবেল ফাইলগুলো চেক এবং ফিক্স করার ফাংশন।"""
    incorrect_files = []

    for label_file in os.listdir(label_dir):
        if label_file.endswith('.txt'):
            file_path = os.path.join(label_dir, label_file)
            with open(file_path, 'r') as file:
                lines = file.readlines()

            fixed_lines = []
            has_issue = False

            for line in lines:
                parts = line.strip().split()
                if len(parts) == 5:  # সঠিক YOLO ফরম্যাট চেক করা
                    fixed_lines.append(line)
                else:
                    has_issue = True

            if has_issue:
                incorrect_files.append(label_file)
                if fixed_lines:
                    # শুধুমাত্র সঠিক লাইনগুলো সংরক্ষণ করা
                    with open(file_path, 'w') as file:
                        file.writelines(fixed_lines)
                else:
                    # যদি সব লাইন ভুল হয়, ফাইলটি মুছে ফেলা
                    os.remove(file_path)
                    print(f"Invalid label file removed: {label_file}")

    return incorrect_files

# Train এবং Validation লেবেল ফাইল চেক এবং ফিক্স করা
train_issues = check_and_fix_labels(train_label_path)
val_issues = check_and_fix_labels(val_label_path)

# সমস্যার রিপোর্ট
if train_issues:
    print(f"Train লেবেল সমস্যা: {train_issues}")
else:
    print("Train লেবেল ফাইলগুলো সঠিক।")

if val_issues:
    print(f"Validation লেবেল সমস্যা: {val_issues}")
else:
    print("Validation লেবেল ফাইলগুলো সঠিক।")
```
